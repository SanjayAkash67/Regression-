```python
import pandas as pd
```

```python

df = pd.read_csv("datasets/ice_cream_sale_temp/IceCreamData.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
X=df[["Temperature"]]
y=df["Revenue"]
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=10, include_bias=False)
X_poly = poly.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import Lasso, Ridge
ridge_model = Ridge(alpha=0.1)
ridge_model.fit(X_train, y_train)
y_pred_ridge = ridge_model.predict(X_test)

lasso_model = Lasso(alpha=0.1,max_iter=10000)
lasso_model.fit(X_train, y_train)
y_pred_lasso = lasso_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

maer = mean_absolute_error(y_test, y_pred_ridge)
mser = mean_squared_error(y_test, y_pred_ridge)
rmser = root_mean_squared_error(y_test, y_pred_ridge)
r2r = r2_score(y_test, y_pred_ridge)

mael = mean_absolute_error(y_test, y_pred_lasso)
msel = mean_squared_error(y_test, y_pred_lasso)
rmsel = root_mean_squared_error(y_test, y_pred_lasso)
r2l = r2_score(y_test, y_pred_lasso)

```

```python
print("Ridge Regression Metrics:")
print(f"MAE: {maer:.4f}")
print(f"MSE: {mser:.4f}")
print(f"RMSE: {rmser:.4f}")
print(f"R²: {r2r:.4f}")

print("\nLasso Regression Metrics:")
print(f"MAE: {mael:.4f}")
print(f"MSE: {msel:.4f}")
print(f"RMSE: {rmsel:.4f}")
print(f"R²: {r2l:.4f}")
```
