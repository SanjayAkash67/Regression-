```python
import pandas as pd
df = pd.read_csv("datasets/ice_cream_sale_temp/IceCreamData.csv")
```

```python
X = df[['Temperature']]
y = df['Revenue']
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=10, include_bias=False)
X_poly = poly.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_poly, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import ElasticNet, SGDRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error

# Note: SGDRegressor is very sensitive to feature scales, so we scale X_train and X_test first!
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 1. Train ElasticNet (Combines L1 and L2 penalties via l1_ratio)
elastic_net = ElasticNet(alpha=1.0, l1_ratio=0.5, max_iter=10000)
elastic_net.fit(X_train, y_train)
y_pred_en = elastic_net.predict(X_test)

# 2. Train SGD Regressor (Gradient Descent optimization)
sgd_model = SGDRegressor(max_iter=1000, tol=1e-3, random_state=42)
sgd_model.fit(X_train_scaled, y_train)
y_pred_sgd = sgd_model.predict(X_test_scaled)

# 3. Evaluate ElasticNet
print("--- ElasticNet Regression ---")
print(f"ElasticNet MAE  : {mean_absolute_error(y_test, y_pred_en):.4f}")
print(f"ElasticNet RMSE : {mean_squared_error(y_test, y_pred_en):.4f}")
print(f"ElasticNet RMSE : {root_mean_squared_error(y_test, y_pred_en):.4f}")
print(f"ElasticNet R²   : {r2_score(y_test, y_pred_en):.4f}")

# 4. Evaluate SGD Regressor
print("\n--- SGD Regressor ---")
print(f"SGD MAE  : {mean_absolute_error(y_test, y_pred_sgd):.4f}")
print(f"SGD RMSE : {mean_squared_error(y_test, y_pred_sgd):.4f}")
print(f"SGD RMSE : {root_mean_squared_error(y_test, y_pred_sgd):.4f}")
print(f"SGD R²   : {r2_score(y_test, y_pred_sgd):.4f}")
```
